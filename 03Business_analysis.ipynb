```python
import pandas as pd
```

```python
import numpy as np
```

```python
df = pd.read_csv('../Data/Cleaned_Data/airbnb_cleaned.csv')
```

```python
df.head()
```

```python
# Which neighbourhood earns the most on average, and which has the most listings?
df.groupby('neighbourhood')['price'].agg(['mean','count']).sort_values('mean',ascending = False)
```

```python
# which room type performs best  - price vs demand?
df.groupby('room_type').agg(avg_price=('price','mean'),avg_reviews=('number_of_reviews','mean'))
```

```python
# Does a faster host response rate lead to a better rating?
df['host_response_rate'] = df['host_response_rate'].astype(str).str.replace("%","").astype(float)
df[['host_response_rate','rating']].corr()
```

```python
# Does a stricter cancellation policy scare guests away?
df.groupby('cancellation_policy').agg(avg_reviews=('number_of_reviews','mean'),avg_price=('price','mean'))
```
